STEP 2 MISSING VALUES ANALYSIS

MISSING VALUES COUNT

```python
count = df.isnull().sum()
print("NO of Missing Values are:")
print(count)
```

MISSING VALUES PERCENTAGE

```python
PERCT = (df.isnull().sum()/len(df))*100
print("percentage of missing values are:")
print(F"{PERCT}")
```

ROWS WITH MISSING VALUES

```python
any_rows_with_missing_values = df.isnull().any(axis=1).sum()
print("Any rows with missing values are :{any_rows_with_missing_values}")
```
