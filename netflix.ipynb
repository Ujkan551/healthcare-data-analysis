THERE ARE TWO MUCH NULL VALUES IN COLUMNS 

IF we see the sample size of 50 rows, there are null values in almost every other row so if i remove all these rows nothing will be left for analysis

instead of removing every other row i will fill these data 

For categorical data i cannot use mode as those missing data contains name as it contains cast,director,country i cannot fill these data with mode value this can make my analysis fake and fraud

instead i will fill them with different words


```python
def fill_missing_data(df):
    for col in df.columns:
        if df[col].isnull().any():
            if df[col].dtype == 'object':
                if df["director"].isnull().any():
                    df["director"] = df["director"].fillna("unkown director")
                if df["cast"].isnull().any():
                    df["cast"] = df["cast"].fillna("cast not known")
                if df["country"].isnull().any():
                    df["country"] = df["country"].fillna("country data not filled")
            else:
                # Fill with the mean for numerical data
                df[col] = df[col].fillna(df[col].mean())
    return df
```

Here i made a function first it iterates through each columns then i check which -which col contains null values after that i split cols in two types first categorical columns then numerical columns numerical cols are filled with mean value(most appropiate for this dataset)
